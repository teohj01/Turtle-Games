```python
import pandas as pd
import statsmodels.api as sm 
from statsmodels.formula.api import ols
```

```python
# Load the dataset
df = pd.read_csv("data/cleaned_turtle_reviews.csv")
df.head()
```

## Linear Regression

### Simple Linear Regression

`remuneration` vs `loyalty_points`

```python
# Define independent variable
y = df['loyalty_points']

# Define dependent variable
x = df['remuneration']

# Add constant to x for intercept
x = sm.add_constant(x)

# Fit the linear regression model
model = sm.OLS(y, x).fit()

# Print the summary of the regression model
print(model.summary())
```

```python
# Extract the estimated parameters and standard errors
params = model.params
std_err = model.bse
intercept = params.get('const', 0)  # if constant is present

# Build the regression equation from all coefficients
terms = " + ".join("{:.2f} * {}".format(params[var], var)
                   for var in params.index if var != 'const')
print("Regression equation: loyalty_points = {} + {:.2f}".format(terms, intercept))

# View the output.
print("\nRegression Table")
print("-" * 50)
print("{:<18}{:>14}{:>16}".format("Variable", "Coefficient", "Standard Error"))
print("-" * 50)
for var in params.index:
    name = "Intercept" if var == 'const' else var
    print("{:<18}{:>14.2f}{:>16.2f}".format(name, params[var], std_err[var]))
```

#### Remuneration vs Loyalty Points
`remuneration` is a statistically significant predictor of `loyalty_points` (p < 0.001). 

The slope of 34.19 means loyalty points increase by 34.19 for every one-unit increase in remuneration. 

The R² value of 0.38, showing a moderate fit but the spread of data suggests other factors also affecting loyalty points.

### Multivariate Linear Regression

`loyalty_points` = intercept + b1 * `remuneration` + b2 * `spending_score`

```python
# Define independent variable
y = df['loyalty_points']

# Define dependent variables
x = df[['remuneration', 'spending_score']]

# Add constant to x for intercept
x = sm.add_constant(x)

# Fit the linear regression model
model = sm.OLS(y, x).fit()

# Print the summary of the regression model
print(model.summary())
```

```python
# Extract the estimated parameters and standard errors
params = model.params
std_err = model.bse
intercept = params.get('const', 0)  # if constant is present

# Build the regression equation from all coefficients
terms = " + ".join("{:.2f} * {}".format(params[var], var)
                   for var in params.index if var != 'const')
print("Regression equation: loyalty_points = {} + {:.2f}".format(terms, intercept))

# View the output.
print("\nRegression Table")
print("-" * 50)
print("{:<18}{:>14}{:>16}".format("Variable", "Coefficient", "Standard Error"))
print("-" * 50)
for var in params.index:
    name = "Intercept" if var == 'const' else var
    print("{:<18}{:>14.2f}{:>16.2f}".format(name, params[var], std_err[var]))
```

For each one-point increase in spending score, loyalty points increase by about 32.89, holding remuneration constant.

For every one-unit increase in remuneration, loyalty points increase by about 33.97, holding spending score constant.

The R² value of 0.827 indicates a strong fit, with remuneration and spending score together explaining around 83% of the variation in loyalty points.
